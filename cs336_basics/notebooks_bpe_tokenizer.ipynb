```python
import regex as re

PAT = r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+"""
re.findall(PAT, "some text that i'll pre-tokenize")
```

```python
example_text = """
low low low low low
lower lower widest widest widest
newest newest newest newest newest newest
"""
pre_tokens = re.findall(PAT, example_text)
print(pre_tokens)
```

```python
from collections import defaultdict
pre_tkn_count = defaultdict(int)
for tkn in pre_tokens:
    pre_tkn_count[tuple(tkn)] += 1
print(pre_tkn_count)
```

```python
tokens = defaultdict(int)
for tkn in pre_tkn_count:
    for i in range(len(tkn)-1):
        tokens[tkn[i]+tkn[i+1]] += 1 * pre_tkn_count[tkn]

print(tokens)
```
